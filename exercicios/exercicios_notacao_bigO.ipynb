### Identificação da complexidade

**(1)** Qual é a complexidade de `consultar_primeiro`? Aumentar o tamanho da lista altera a quantidade de operações?

```python
def consultar_primeiro(valores):
    return valores[0]
```

**(2)** Determine a complexidade de `somar_valores` em função da quantidade de elementos da lista.

```python
def somar_valores(valores):
    total = 0
    for valor in valores:
        total += valor
    return total
```

**(3)** O código abaixo percorre a lista duas vezes, em laços separados. Qual é a complexidade total? Explique por que os custos dos laços são somados.

```python
for valor in valores:
    print(valor)

for valor in valores:
    print(valor)
```

**(4)** Identifique a complexidade dos laços aninhados. Quantas vezes, aproximadamente, a instrução interna é executada em função de `n`?

```python
for i in range(n):
    for j in range(n):
        print(i, j)
```

**(5)** A cada repetição, o valor de `tamanho` é dividido por dois. Qual é a complexidade do laço?

```python
while tamanho > 1:
    tamanho //= 2
```

**(6)** O laço externo executa `n` vezes. Como cresce o número de repetições do laço interno?

```python
for i in range(n):
    j = 1
    while j < n:
        j *= 2
```

**(7)** Neste trecho, os laços percorrem entradas de tamanhos possivelmente diferentes, `n` e `m`. Expresse a complexidade usando as duas variáveis.

```python
for i in range(n):
    for j in range(m):
        print(i, j)
```

**(8)** Qual é a complexidade do trecho com três laços aninhados, cada um executado `n` vezes?

```python
for i in range(n):
    for j in range(n):
        for k in range(n):
            print(i, j, k)
```

**(9)** Simplifique a complexidade total. Qual parcela domina quando `n` cresce?

```python
for valor in valores:
    print(valor)          # O(n)

for i in range(n):
    for j in range(n):
        print(i, j)        # O(n²)

print("fim")                   # O(1)
```

### Comparação entre algoritmos

**(10)** As duas funções calculam a soma dos números de `1` até `n`. Determine a complexidade de cada uma e compare como o tempo de execução cresce para valores grandes de `n`.

```python
def soma_por_repeticao(n):
    total = 0
    for numero in range(1, n + 1):
        total += numero
    return total

def soma_por_formula(n):
    return n * (n + 1) // 2
```

**(11)** As funções abaixo encontram o menor valor da lista. Compare o tempo de execução e explique por que ordenar a lista inteira faz mais trabalho do que percorrê-la uma vez.

```python
def menor_por_percurso(valores):
    menor = valores[0]
    for valor in valores:
        if valor < menor:
            menor = valor
    return menor

def menor_ordenando(valores):
    return sorted(valores)[0]
```

**(12)** As duas funções verificam se há valores repetidos. Compare o tempo de execução no caso em que não há valores repetidos e o algoritmo de pares precisa comparar todos eles. Para o algoritmo com `set`, considere consultas em tempo `O(1)` em média. Qual algoritmo usa memória adicional proporcional ao tamanho da lista?

```python
def repetidos_comparando_pares(valores):
    for i in range(len(valores)):
        for j in range(i + 1, len(valores)):
            if valores[i] == valores[j]:
                return True
    return False

def repetidos_com_conjunto(valores):
    vistos = set()
    for valor in valores:
        if valor in vistos:
            return True
        vistos.add(valor)
    return False
```
