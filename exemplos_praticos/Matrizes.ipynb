# Matrizes

```python
import numpy as np
```

criando a matriz

```python
matriz = np.array([[2, 3, 1],
                   [4, 5, 5]] )
matriz
```

matriz com valores ausentes 

```python
matriz = np.array([[2, 3, 1],
                   [4, 5, None]] )
matriz
```

verificar o shape da matriz

```python
matriz.shape
```

acessar linha da matriz


```python
matriz[0,:]
```

Acessar coluna

```python
matriz[:, 0]
```

acessar um elemento da matriz


```python
matriz[0][2]
```

```python
matriz[1][2]
```

```python
for i in range(matriz.shape[0]):
  print(matriz[i])
  for j in range(matriz.shape[1]):
    print(matriz[i][j])
```
