# Exemplo 1

Ler 5 notas e informar a média


## For

```python
soma = 0
for _ in range(1, 6):
  nota = float(input('Digite a nota '))
  soma += nota
print(f"A média é: {soma / 5}")
```

## While

```python
soma = 0
numero = 1
while numero <= 5:
  nota = float(input('Digite a nota:'))
  soma += nota
  numero += 1
print(f"A média é: {soma / 5}")
print(f"Qtd de notas é: {numero-1}")
```

# Exemplo 2

Imprimir a tabuada do número 3 


## For

```python
for i in range(1, 11):
  print(f"3 x {i} = {3*i}")
```

## While

```python
numero = 1
while numero <= 10:
  print(f"3 x {numero} = {3*numero}")
  numero += 1
```

### Break 

Usado para interromper uma repetição

```python
for numero in range(1, 11):
    if numero == 6:
        break
    else:
        print(numero)
```

### Continue 

Usado para pular uma repetição de acordo com uma condição

```python
for numero in range(1, 11):
    if numero % 2 == 0:
        continue
    else:
        print(numero)
```
