# Estruturas de repetição - for

Suponha o caso que desejamos exibir na tela a seguência de 1 até 5:

```python
print(1)
print(2)
print(3)
print(4)
print(5)
```

Para fazer isso de forma mais eficiente, utilizamos o operador de repetição *for*: 

Em portugol, seria algo assim: Para número entre 1 e 5 então ...

```python
for numero in range(1, 11):
  print(numero, end=" ")
```

podemos fazer o intervalo invertido: 

```python
for numero in range(5, 0, -1):
  print(numero)
```

```python
for i in range(0,5):
  print(i)
  print('---')
  for j in range(0,3):
    print(j)
  print()
```

### `for` com strings

Uma string é uma sequência de caracteres. Por isso, o `for` pode percorrer o texto e acessar uma letra de cada vez.

No exemplo a seguir, a variável `letra` recebe primeiro `s`, depois `o`, depois `r` e assim por diante, até percorrer toda a palavra.

```python
palavra = 'sorvete'
for letra in palavra:
  print(letra)
```

```python
palavra = 'sorvete'
for letra in palavra:
  if letra == 'v':
    print('Achou a letra v')
```

```python
quantidade = int(input("Quantas notas serão informadas? "))
soma = 0

for indice in range(quantidade):
  nota = float(input(f"Nota {indice+1}: "))
  soma = soma + nota

media = soma / quantidade

if media >= 7:
  situacao = "aprovado"
else:
  situacao = "reprovado"

print(f"Média: {media:.1f}")
print(f"Situação: {situacao}")
```
